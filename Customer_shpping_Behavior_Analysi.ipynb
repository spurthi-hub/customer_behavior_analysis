```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
labels=['Young Adult','Adult','Middle-Aged','Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
frequency_mapping = {
'Fortnightly': 14,
'Weekly': 7,
'Monthly': 30,
'Quarterly': 90,
'Bi-Weekly': 14,
'Annually': 365,
'Every 3 Months' : 90
}
df[ 'purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
import urllib.parse
from sqlalchemy import create_engine

# Step 1: Securely handle credentials and format the connection string
username = "postgres"  # Crucial: the system level master user is 'postgres'
raw_password = "Gadag@1234"
host = "localhost"
port = "5432"
database = "customer_behavior"

# This cleanly converts the '@' symbol so SQLAlchemy doesn't confuse it with the host delimiter
safe_password = urllib.parse.quote_plus(raw_password)

# Note the removal of the accidental space character inside the connection string structure
engine = create_engine(f"postgresql+psycopg2://{username}:{safe_password}@{host}:{port}/{database}")

# Step 2: Load clean DataFrame into PostgreSQL
table_name = "customer"
df.to_sql(table_name, engine, if_exists="replace", index=False)

print(f"Data successfully loaded into table '{table_name}' in database '{database}'.")
```
